4.Most Purchased Products

```python
from matplotlib import colors
data["Product Category"].value_counts().plot(kind="barh",figsize=(10,5),color="green")
plt.legend()
plt.show()
```

```python
top_products=data.groupby('Product Category')["Quantity"].sum().sort_values(ascending=False).plot(kind="bar")
print(top_products)
```

5.Group customers based on spending.

```python
cus_spending=data.groupby('Customer ID')['Returns'].value_counts()
```

```python
print(cus_spending)
```

6.churn values counts

```python
data["Churn"].value_counts().plot(kind="pie")
plt.xlabel("Churn value Count")
plt.legend()
plt.show()
```

7.Making the group based on the gender with churn

```python
data.groupby("Churn")["Gender"].value_counts()
```

```python
data.groupby("Churn")["Gender"].value_counts().plot(kind="bar",color={"blue"})
plt.legend()
plt.show()
```

8.Finding the payment modes
